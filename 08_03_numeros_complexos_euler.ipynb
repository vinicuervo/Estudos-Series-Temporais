## Seção 8.3 — Números complexos e fórmula de Euler

### 1. Intuição: por que representar ondas com números complexos?

Nas Seções 8.1 e 8.2, uma oscilação era descrita por $R\cos(\omega t+\phi)$. Multiplicar, somar e deslocar muitas senoides por métodos trigonométricos torna-se trabalhoso. No plano complexo, a mesma oscilação pode ser associada a um vetor que **gira** em torno da origem. A parte real desse vetor é um cosseno; a imaginária, um seno. Rotacionar o vetor equivale a multiplicar números complexos, o que simplifica expressões e demonstrações.

### 2. Definição e operações fundamentais

A unidade imaginária $i$ satisfaz $i^2=-1$. Um número complexo é $z=a+ib$, com $a,b\in\mathbb R$, parte real $\operatorname{Re}(z)=a$, parte imaginária $\operatorname{Im}(z)=b$ e conjugado $\bar z=a-ib$.

**Soma:** $(a+ib)+(c+id)=(a+c)+i(b+d)$.

**Produto:**

$$
(a+ib)(c+id)=ac+iad+ibc+i^2bd=(ac-bd)+i(ad+bc).
$$

**Módulo:** $|z|=\sqrt{a^2+b^2}$; logo $z\bar z=(a+ib)(a-ib)=a^2+b^2=|z|^2$. Se $z\neq0$, então

$$
\frac1z=\frac{\bar z}{|z|^2}=\frac{a-ib}{a^2+b^2}.
$$

**Exemplo numérico:** $z_1=2+i$ e $z_2=1-2i$ fornecem $z_1z_2=4-3i$, $|z_1|=\sqrt5$, $|z_2|=\sqrt5$ e $|z_1z_2|=5$.

### 3. Forma polar e interpretação geométrica

Para $z\neq0$, escrevemos $z=r(\cos\theta+i\sin\theta)$, onde $r=|z|$ e $\theta=\operatorname{atan2}(b,a)$ é um argumento de $z$. Os argumentos diferem por múltiplos de $2\pi$. O uso de `atan2` é importante: uma tangente $b/a$ isolada não determina o quadrante.

Para $z_1=r_1(\cos\theta_1+i\sin\theta_1)$ e $z_2=r_2(\cos\theta_2+i\sin\theta_2)$, as fórmulas trigonométricas de adição dão

$$
z_1z_2=r_1r_2[\cos(\theta_1+\theta_2)+i\sin(\theta_1+\theta_2)].
$$

Portanto, **módulos se multiplicam e argumentos se somam**, módulo $2\pi$.

### 4. Demonstração da fórmula de Euler

As séries de Taylor, convergentes para todo número complexo $w$, são

$$
e^w=\sum_{n=0}^{\infty}\frac{w^n}{n!},\qquad
\cos x=\sum_{k=0}^{\infty}\frac{(-1)^k x^{2k}}{(2k)!},\qquad
\sin x=\sum_{k=0}^{\infty}\frac{(-1)^k x^{2k+1}}{(2k+1)!}.
$$

Substituindo $w=ix$ e separando potências pares e ímpares — operação justificada pela convergência absoluta — obtemos

$$
\begin{aligned}
e^{ix}&=\sum_{k=0}^{\infty}\frac{i^{2k}x^{2k}}{(2k)!}
+\sum_{k=0}^{\infty}\frac{i^{2k+1}x^{2k+1}}{(2k+1)!}\\
&=\sum_{k=0}^{\infty}\frac{(-1)^k x^{2k}}{(2k)!}
+i\sum_{k=0}^{\infty}\frac{(-1)^k x^{2k+1}}{(2k+1)!}\\
&=\cos x+i\sin x.
\end{aligned}
$$

Logo,

$$\boxed{e^{ix}=\cos x+i\sin x.}$$

Substituir $x$ por $-x$ produz $e^{-ix}=\cos x-i\sin x$, pois cosseno é par e seno é ímpar. **Somando e subtraindo** essas identidades:

$$
\boxed{\cos x=\frac{e^{ix}+e^{-ix}}2},\qquad
\boxed{\sin x=\frac{e^{ix}-e^{-ix}}{2i}}.
$$

Observe que $|e^{ix}|=\sqrt{\cos^2x+\sin^2x}=1$: a exponencial complexa de argumento real percorre o círculo unitário.

### 5. Relação com a série temporal harmônica

Morettin e Toloi (§13.2, eq. 13.1) consideram

$$Z_t=\mu+R\cos(\omega t+\phi)+\varepsilon_t.$$

Pela identidade de Euler, a parte determinística é

$$R\cos(\omega t+\phi)=\operatorname{Re}\big(Re^{i\phi}e^{i\omega t}\big).$$

Definindo $C=Re^{i\phi}$, precisamos observar **a convenção de sinais:** na eq. (13.2) do livro, $A=R\cos\phi$ e $B=-R\sin\phi$, portanto $C=A-iB=R(\cos\phi+i\sin\phi)$; assim,

$$\operatorname{Re}\big((A-iB)e^{i\omega t}\big)=A\cos(\omega t)+B\sin(\omega t).$$

Essa igualdade pode ser confirmada expandindo o produto: $(A-iB)(\cos\omega t+i\sin\omega t)$ tem parte real $A\cos\omega t+B\sin\omega t$.

**Exemplo:** para $R=3$, $\phi=-\pi/6$, $\omega=\pi/4$, temos $A=3\sqrt3/2$, $B=3/2$ e $C=3e^{-i\pi/6}=3\sqrt3/2-1{,}5i=A-iB$.

### 6. Exemplo computacional: círculo unitário e projeções

Antes da implementação: NumPy armazena números complexos nativamente (`1j` representa $i$); `np.exp` calcula a exponencial; `np.real`, `np.imag` e `np.abs` extraem parte real, imaginária e módulo. Matplotlib permite comparar as projeções.

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 600)
z = np.exp(1j*theta)
assert np.allclose(z.real, np.cos(theta))
assert np.allclose(z.imag, np.sin(theta))
assert np.allclose(np.abs(z), 1)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(z.real, z.imag)
axes[0].set(xlabel='Parte real', ylabel='Parte imaginária', title='Euler: círculo unitário')
axes[0].set_aspect('equal', adjustable='box')
axes[0].grid(alpha=.25)
axes[1].plot(theta, z.real, label='Re = cosseno')
axes[1].plot(theta, z.imag, label='Im = seno')
axes[1].set(xlabel='Ângulo (rad)', ylabel='Projeção', title='Duas projeções da mesma rotação')
axes[1].legend(); axes[1].grid(alpha=.25)
plt.tight_layout(); plt.show()
```

**Interpretação:** a curva da esquerda é circular, mas cada coordenada isoladamente oscila. O gráfico ilustra a identidade; não constitui sua demonstração formal, apresentada na §4.

### 7. Soma de ondas com números complexos

Se as duas ondas têm a **mesma frequência**,

$$R_1\cos(\omega t+\phi_1)+R_2\cos(\omega t+\phi_2)=\operatorname{Re}\big[(R_1e^{i\phi_1}+R_2e^{i\phi_2})e^{i\omega t}\big].$$

O número complexo $C=R_1e^{i\phi_1}+R_2e^{i\phi_2}$ determina a amplitude $|C|$ e a fase $\arg C$ resultantes. Isso **não** permite reduzir, em geral, oscilações de frequências distintas a uma única senoide.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 12, 800)
omega = 2*np.pi/6
R1, phi1 = 2.0, 0.0
R2, phi2 = 1.0, np.pi/2
C = R1*np.exp(1j*phi1) + R2*np.exp(1j*phi2)
y_direto = R1*np.cos(omega*t+phi1) + R2*np.cos(omega*t+phi2)
y_complexo = (C*np.exp(1j*omega*t)).real
assert np.allclose(y_direto, y_complexo)
print(f'Amplitude resultante: {abs(C):.6f}; fase (rad): {np.angle(C):.6f}')
fig, ax = plt.subplots(figsize=(9,3.5))
ax.plot(t, y_direto, label='Soma direta')
ax.plot(t, y_complexo, '--', label='Representação complexa')
ax.set(xlabel='Tempo', ylabel='Sinal', title='Soma de senoides de mesma frequência')
ax.legend(); ax.grid(alpha=.25); plt.tight_layout(); plt.show()
```

**Interpretação:** os gráficos coincidem; neste exemplo a amplitude resultante é $\sqrt5$, não $3$, porque as duas ondas possuem fases diferentes.

### 8. Limitações e cuidados

- A identidade de Euler é uma igualdade exata, mas os gráficos e as comparações `allclose` são verificações numéricas sujeitas a arredondamento.
- A fase fica indefinida quando a amplitude é zero; o argumento complexo não é único.
- A representação complexa **não transforma uma série real em fenômeno físico complexo**: é uma ferramenta algébrica.
- A expansão em exponenciais prepara Fourier, mas não garante, por si só, que qualquer função admita uma série convergente ponto a ponto.
