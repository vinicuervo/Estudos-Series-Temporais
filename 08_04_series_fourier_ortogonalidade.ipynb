## Seção 8.4 — Séries de Fourier e ortogonalidade

**Objetivo:** compreender como uma função periódica pode ser representada por senos e cossenos, deduzir as fórmulas dos coeficientes por ortogonalidade e conectar a série contínua à análise harmônica **discreta** utilizada em séries temporais.

(13.26)–(13.28), análise de Fourier discreta, coeficientes e partição da variância. **Séries de Fourier contínuas, provas de ortogonalidade integral e condições de convergência são complementações matemáticas** para construir os pré-requisitos da DFT.

### 1. Intuição: representar formas complexas com ondas simples

Um sinal periódico pode ser complicado mesmo quando seu padrão se repete exatamente. A ideia é somar uma onda de frequência fundamental, outras de frequência duas, três ou mais vezes maior e uma constante. A escolha dos pesos permite aproximar a forma observada. Em uma amostra temporal, Morettin e Toloi chamam essa representação de **análise harmônica** (§13.4).

### 2. Frequência fundamental e harmônicos

Uma função $f$ tem período $T>0$ se $f(t+T)=f(t)$ para todo $t$ de seu domínio. Fixado esse período, a frequência angular fundamental é $\omega_0=2\pi/T$ e o harmônico $n$ tem frequência $n\omega_0$.

A expansão trigonométrica formal é

$$
\boxed{f(t)\sim\frac{a_0}{2}+\sum_{n=1}^{\infty}\left[a_n\cos(n\omega_0 t)+b_n\sin(n\omega_0 t)\right].}
$$

O símbolo $\sim$ indica uma expansão proposta; igualdade pontual ou convergência em norma exige condições adicionais. O termo $a_0/2$ descreve o nível médio; $a_n$ e $b_n$ medem as contribuições em cada harmônico.

### 3. Produto interno de funções e ortogonalidade

Para funções reais integráveis em $[0,T]$, definimos

$$\langle f,g\rangle=\int_0^T f(t)g(t)\,dt.$$

Dizemos que $f$ e $g$ são **ortogonais** se $\langle f,g\rangle=0$. É análogo ao produto escalar de vetores perpendicularizados: a integral mede a sobreposição assinada entre duas funções.

Para inteiros $m,n\ge1$,

$$
\begin{aligned}
\int_0^T\cos(n\omega_0t)\cos(m\omega_0t)\,dt&=\frac{T}{2}\delta_{nm},\\
\int_0^T\sin(n\omega_0t)\sin(m\omega_0t)\,dt&=\frac{T}{2}\delta_{nm},\\
\int_0^T\sin(n\omega_0t)\cos(m\omega_0t)\,dt&=0,
\end{aligned}
$$

onde $\delta_{nm}=1$ se $n=m$ e $0$ caso contrário. A constante $1$ também é ortogonal a todos os senos e cossenos de ordem positiva: $\int_0^T\sin(n\omega_0t)dt=\int_0^T\cos(n\omega_0t)dt=0$.

#### Demonstração detalhada

Use a identidade produto–soma:

$$2\cos u\cos v=\cos(u-v)+\cos(u+v).$$

Se $n\ne m$, a integral de cada termo é zero, pois

$$\int_0^T\cos(k\omega_0t)\,dt=\left[\frac{\sin(k\omega_0t)}{k\omega_0}\right]_0^T=0,\qquad k\in\mathbb Z\setminus\{0\}.$$

Se $n=m$, $2\cos^2(n\omega_0t)=1+\cos(2n\omega_0t)$, cuja integral é $T$. Assim a integral de $\cos^2$ é $T/2$. O mesmo raciocínio, usando $2\sin u\sin v=\cos(u-v)-\cos(u+v)$ e $2\sin u\cos v=\sin(u+v)+\sin(u-v)$, demonstra as outras igualdades. **Esse é o motivo matemático que permite recuperar cada coeficiente separadamente.**

### 4. Dedução dos coeficientes contínuos

Considere uma soma **finita**

$$S_K(t)=\frac{a_0}{2}+\sum_{n=1}^K\left[a_n\cos(n\omega_0 t)+b_n\sin(n\omega_0t)\right].$$

Integrando no intervalo completo e usando ortogonalidade,

$$\int_0^T S_K(t)\,dt=\frac{a_0T}{2}\implies a_0=\frac2T\int_0^T S_K(t)\,dt.$$

Agora multiplicamos $S_K$ por $\cos(m\omega_0t)$, $1\le m\le K$, e integramos:

$$\int_0^T S_K(t)\cos(m\omega_0t)\,dt=\frac{T}{2}a_m,$$

pois todos os demais produtos possuem integral zero. Portanto,

$$\boxed{a_n=\frac2T\int_0^T f(t)\cos(n\omega_0 t)\,dt},\qquad
\boxed{b_n=\frac2T\int_0^T f(t)\sin(n\omega_0 t)\,dt}.$$

As fórmulas para $f$ geral são os **coeficientes de projeção** de $f$ no espaço gerado pelas funções trigonométricas. Para $f\in L^2([0,T])$ a projeção finita é válida e as somas parciais convergem em norma $L^2$; isso não afirma convergência pontual em todos os instantes.

### 5. Exemplo numérico exato: dois harmônicos

Considere, em $T=2\pi$,

$$f(t)=2+3\cos(t)-\sin(2t).$$

Pela ortogonalidade, $a_0=4$, $a_1=3$, $b_2=-1$, e todos os outros coeficientes são zero. Observe que **o nível médio é $a_0/2=2$**, uma convenção importante.

Seu quadrado médio centrado ao longo de um período é

$$\frac1{2\pi}\int_0^{2\pi}[f(t)-2]^2\,dt=\frac{3^2}{2}+\frac{(-1)^2}{2}=5.$$

Isso ilustra a identidade de Parseval para um polinômio trigonométrico finito, sem necessidade de argumentar sobre a convergência de uma série infinita.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2*np.pi, 5000, endpoint=False)
f = 2 + 3*np.cos(t) - np.sin(2*t)
T = 2*np.pi
# Integração por soma retangular periódica, com espaçamento uniforme dt
D = t[1]-t[0]
a0 = (2/T)*D*np.sum(f)
a = [(2/T)*D*np.sum(f*np.cos(n*t)) for n in range(1,5)]
b = [(2/T)*D*np.sum(f*np.sin(n*t)) for n in range(1,5)]
print('a0 =', a0, 'a =', np.round(a,4), 'b =', np.round(b,4))
assert np.allclose(a0, 4, atol=1e-10)
assert np.allclose(a, [3,0,0,0], atol=1e-10)
assert np.allclose(b, [0,-1,0,0], atol=1e-10)
fig, ax = plt.subplots(figsize=(9,3.5))
ax.plot(t, f, label='f(t) = 2 + 3 cos(t) - sen(2t)')
ax.plot(t, 2 + 3*np.cos(t), '--', label='Sem o 2º harmônico')
ax.set(xlabel='t (rad)', ylabel='f(t)', title='Composição de dois harmônicos')
ax.grid(alpha=.25); ax.legend(); plt.tight_layout(); plt.show()
```

**Interpretação:** o segundo harmônico altera a forma da onda sem mudar seu nível médio. A soma retangular é uma aproximação numérica; neste exemplo, a grade uniforme e as frequências inteiras fazem as somas discretas reproduzirem as integrais praticamente até a precisão da máquina.

### 6. Exemplo com descontinuidade e fenômeno de Gibbs

Considere a onda quadrada periódica de período $2\pi$ definida por $f(t)=1$ para $0<t<\pi$ e $f(t)=-1$ para $-\pi<t<0$. É ímpar: $a_0=a_n=0$. Para $n\ge1$,

$$
\begin{aligned}
b_n&=\frac1\pi\int_{-\pi}^{\pi}f(t)\sin(nt)\,dt
=\frac2\pi\int_0^\pi\sin(nt)\,dt\\
&=\frac{2}{\pi n}[1-(-1)^n]
=\begin{cases}\frac4{\pi n},& n\ \text{ímpar},\\0,&n\ \text{par}.\end{cases}
\end{aligned}
$$

Assim,

$$f(t)\sim\frac4\pi\left[\sin t+\frac{\sin 3t}{3}+\frac{\sin 5t}{5}+\cdots\right].$$

Nos pontos de descontinuidade, a soma de Fourier converge para a média dos limites laterais, isto é, zero, sob as condições clássicas de Dirichlet. Perto das descontinuidades aparecem oscilações (*fenômeno de Gibbs*): acrescentar harmônicos estreita a região da oscilação, mas não elimina seu sobrepasso máximo.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-np.pi, np.pi, 2000, endpoint=False)
f = np.where(t<0, -1.0, 1.0)
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(t, f, color='black', linewidth=2, label='Onda quadrada (valor por convenção em 0)')
for N in [1, 5, 25]:
    indices = np.arange(1, N+1, 2)
    s = (4/np.pi)*np.sum(np.sin(indices[:,None]*t[None,:])/indices[:,None], axis=0)
    ax.plot(t, s, label=f'Harmônicos ímpares até n={N}')
ax.set(xlabel='t (rad)', ylabel='Amplitude', title='Somas parciais de Fourier e fenômeno de Gibbs')
ax.legend(); ax.grid(alpha=.2); plt.tight_layout(); plt.show()
```

**Interpretação:** longe das descontinuidades, a aproximação melhora com mais harmônicos. Nos saltos, não se espera convergência ao valor arbitrariamente atribuído a $f(0)$.

### 7. Ponte indispensável: do contínuo ao discreto (livro, §13.4)

A fórmula anterior integra **uma função ao longo de um período contínuo**. Em séries temporais, observamos $z_0,\ldots,z_{N-1}$ em instantes discretos. Para $N$ par, a representação harmônica completa dos $N$ valores é

$$
z_t=a_0+\sum_{j=1}^{N/2-1}\left[a_j\cos\left(\frac{2\pi jt}{N}\right)+b_j\sin\left(\frac{2\pi jt}{N}\right)\right]+a_{N/2}(-1)^t,
$$

em que $a_0=\bar z$, $a_j=\frac2N\sum_{t=0}^{N-1}z_t\cos(2\pi jt/N)$, $b_j=\frac2N\sum_{t=0}^{N-1}z_t\sin(2\pi jt/N)$ e $a_{N/2}=\frac1N\sum_{t=0}^{N-1}(-1)^t z_t$.

Essa é a forma das equações **(13.26)–(13.27), pp. 348–349**, com índices deslocados de $t=1,\ldots,N$ no livro para $t=0,\ldots,N-1$ aqui. A diferença de origem altera as fases, mas não a substância da decomposição.

**Por que os fatores diferem do caso contínuo?** No modelo contínuo, usamos $a_0/2$; no discreto acima, a constante é diretamente $a_0=\bar z$. Além disso, para $N$ par, o termo de Nyquist $j=N/2$ é especial: o seno se anula em todos os pontos amostrados e o cosseno vira $(-1)^t$.

A identidade de ortogonalidade discreta surge da soma geométrica para $k\in\mathbb Z$:

$$\sum_{t=0}^{N-1}e^{2\pi i kt/N}=\begin{cases}N,&k\equiv0\pmod N,\\0,&\text{caso contrário}.\end{cases}$$

Para $k\not\equiv0\pmod N$, defina $q=e^{2\pi ik/N}\ne1$; então $\sum_{t=0}^{N-1}q^t=(1-q^N)/(1-q)=0$, pois $q^N=e^{2\pi ik}=1$. Isso justifica, pela fórmula de Euler, o isolamento dos coeficientes discretos.

### 8. Energia e partição da variância

As funções da base são ortogonais também no produto interno discreto $\langle x,y\rangle=\sum_{t=0}^{N-1}x_ty_t$. Para $N$ par,

$$\boxed{\frac1N\sum_{t=0}^{N-1}(z_t-\bar z)^2=\frac12\sum_{j=1}^{N/2-1}(a_j^2+b_j^2)+a_{N/2}^2.}$$

É a partição da variância apresentada no livro, **eq. (13.28), p. 349**, para a normalização de seus coeficientes. O lado direito quantifica o quanto cada frequência contribui para a variabilidade amostral, tomando o denominador $N$ (não $N-1$).

```python
import numpy as np
import matplotlib.pyplot as plt

N = 48
t = np.arange(N)
z = 4 + 2.0*np.cos(2*np.pi*4*t/N) - 1.5*np.sin(2*np.pi*9*t/N) + 0.7*(-1)**t
bar_z = np.mean(z)
a = np.array([(2/N)*np.sum(z*np.cos(2*np.pi*j*t/N)) for j in range(1,N//2)])
b = np.array([(2/N)*np.sum(z*np.sin(2*np.pi*j*t/N)) for j in range(1,N//2)])
a_nyquist = np.mean(z*(-1)**t)
reconstrucao = bar_z + np.sum(
    a[:,None]*np.cos(2*np.pi*np.arange(1,N//2)[:,None]*t/N) +
    b[:,None]*np.sin(2*np.pi*np.arange(1,N//2)[:,None]*t/N), axis=0
) + a_nyquist*(-1)**t
var_nivel = np.mean((z-bar_z)**2)
var_harmonicos = 0.5*np.sum(a*a+b*b) + a_nyquist**2
assert np.allclose(z, reconstrucao, atol=1e-11)
assert np.allclose(var_nivel, var_harmonicos, atol=1e-11)
print(f'Média = {bar_z:.4f}; variância temporal = {var_nivel:.4f}; soma espectral = {var_harmonicos:.4f}')
fig, ax = plt.subplots(figsize=(9,3.5))
ax.stem(np.arange(1,N//2), np.sqrt(a*a+b*b), basefmt=' ')
ax.set(xlabel='Índice harmônico j', ylabel='Amplitude R_j', title='Harmônicos do sinal amostrado (exceto Nyquist)')
ax.grid(alpha=.25); plt.tight_layout(); plt.show()
```

**Interpretação:** a decomposição recupera os harmônicos $j=4$ e $j=9$, enquanto o termo Nyquist é contabilizado separadamente. A reconstrução é exata dentro do arredondamento numérico, pois foram usados **todos** os harmônicos da base discreta.

### 9. Limitações e distinções metodológicas

- Para $f\in L^2$, as somas parciais convergem em norma quadrática, mas convergência pontual pode exigir hipóteses adicionais; uma função integrável qualquer não garante automaticamente convergência pontual da série.
- **Gibbs** é uma consequência de descontinuidades; aumentar os termos não elimina totalmente a oscilação máxima perto do salto.
- A decomposição de $N$ pontos descreve esses pontos, mas não determina unicamente um sinal contínuo entre amostras: existe **aliasing**.
- Frequências fora da grade $j/N$ podem produzir *vazamento espectral*; veremos isso nos módulos de DFT e análise espectral.
- A identidade da variância acima é para dados finitos e denominador $N$, não uma afirmação sobre a variância populacional de um processo aleatório sem hipóteses adicionais.
