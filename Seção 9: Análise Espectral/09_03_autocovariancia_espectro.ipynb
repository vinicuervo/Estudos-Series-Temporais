# Módulo 9 — Análise Espectral
## Seção 9.3 — Relação entre autocovariância e espectro

**Objetivos.** Reconstruir a autocovariância a partir do espectro, compreender os casos em que a representação usa uma medida espectral e não uma densidade, demonstrar como correlações temporais geram concentração espectral e verificar numericamente a dualidade.

**Pré-requisitos.** Módulos 1 e 4; seções 8.3–8.6 e 9.1–9.2. Notação: $X_t$ é real, estacionário de segunda ordem, $\mu=E[X_t]$, $\gamma(h)=\operatorname{Cov}(X_t,X_{t+h})$ e $\lambda$ está em radianos por amostra.

**Fonte principal diretamente conferida.** Morettin e Toloi (2018), *Análise de Séries Temporais: Modelos Lineares Univariados*, v. 1, 3ª ed., **§14.2, pp. 357–362**, eqs. (14.2)–(14.3), e **§14.3, pp. 362–366**, eqs. (14.9)–(14.14). Os exemplos numéricos, a prova explícita de soma geométrica e a implementação Python são complementações didáticas. A notação $F$ é a função de distribuição espectral, não a FFT de uma amostra.

---

### 1. Intuição: duas maneiras de descrever a mesma dependência

A autocovariância responde *quanto duas observações separadas por $h$ instantes se relacionam*. O espectro responde *em quais frequências se distribui a variabilidade de segunda ordem*. Para processos estacionários, esses objetos estão matematicamente associados: o espectro é uma representação da sequência de covariâncias, e não uma fonte independente de informação.

A passagem requer distinguir **densidade espectral**, **medida de distribuição espectral** e **periodograma**, que será introduzido na Seção 9.4. Uma densidade é uma descrição teórica de um processo; um periodograma é uma estatística de uma amostra finita.

### 2. Densidade espectral sob hipótese suficiente

Se $X_t$ é fracamente estacionário, $E(X_t^2)<\infty$ e

$$\sum_{h=-\infty}^{\infty}|\gamma(h)|<\infty,$$

então a série converge absolutamente e uniformemente em $\lambda$ e define

$$\boxed{f(\lambda)=\frac{1}{2\pi}\sum_{h=-\infty}^{\infty}\gamma(h)e^{-i\lambda h}},\qquad -\pi\le\lambda\le\pi.\tag{9.3.1}$$

Como $\gamma(-h)=\gamma(h)$ para processos reais,

$$f(\lambda)=\frac{1}{2\pi}\left[\gamma(0)+2\sum_{h=1}^{\infty}\gamma(h)\cos(h\lambda)\right].\tag{9.3.2}$$

**Observação sobre a hipótese:** somabilidade absoluta é suficiente para uma densidade contínua e para as operações abaixo, mas não é necessária para a existência de uma representação espectral. Não se deve confundir existência de uma *medida* espectral com existência de uma *densidade* usual.

### 3. Demonstração da relação inversa

Para $k\in\mathbb Z$, multiplique (9.3.1) por $e^{ik\lambda}$ e integre:

$$\int_{-\pi}^{\pi}f(\lambda)e^{ik\lambda}\,d\lambda=\frac{1}{2\pi}\sum_{h=-\infty}^{\infty}\gamma(h)\int_{-\pi}^{\pi}e^{i(k-h)\lambda}\,d\lambda.$$

A troca de soma e integral é permitida pela somabilidade absoluta. Para $m\in\mathbb Z$,

$$\int_{-\pi}^{\pi}e^{im\lambda}\,d\lambda=\begin{cases}2\pi,&m=0,\\0,&m\ne0.\end{cases}$$

De fato, se $m\ne0$, $\int e^{im\lambda}\,d\lambda=e^{im\lambda}/(im)$ e $e^{im\pi}=e^{-im\pi}=(-1)^m$; logo a diferença é nula. Resta apenas o termo $h=k$:

$$\boxed{\gamma(k)=\int_{-\pi}^{\pi}e^{ik\lambda}f(\lambda)\,d\lambda=\int_{-\pi}^{\pi}\cos(k\lambda)f(\lambda)\,d\lambda.}\tag{9.3.3}$$

A segunda igualdade decorre da simetria de $f$: o seno é ímpar e sua integral desaparece. Esta é a equação (14.3) do livro. Para $k=0$,

$$\boxed{\gamma(0)=\operatorname{Var}(X_t)=\int_{-\pi}^{\pi}f(\lambda)\,d\lambda.}\tag{9.3.4}$$

**Interpretação.** A autocovariância no atraso zero é a variância total. Em atrasos não nulos, as contribuições de frequências diferentes podem cancelar-se por causa do fator $\cos(k\lambda)$; portanto, uma densidade positiva não implica autocovariâncias positivas.

### 4. Exemplo 1 — Ruído branco

Se $X_t\sim RB(0,\sigma^2)$, $\gamma(0)=\sigma^2$ e $\gamma(h)=0$ para $h\ne0$. Substituindo em (9.3.1):

$$\boxed{f(\lambda)=\sigma^2/(2\pi).}$$

Para $\sigma^2=4$, a altura da curva é $2/\pi$. Sua área é $(2\pi)(2/\pi)=4$. A densidade é constante: nenhuma faixa de frequências é favorecida. Isso NÃO implica que uma realização finita tenha periodograma plano.

### 5. Exemplo 2 — MA(1) e estrutura de altas ou baixas frequências

Considere $X_t=\varepsilon_t+\theta\varepsilon_{t-1}$, com $\varepsilon_t$ ruído branco de variância $\sigma^2$. Por expansão da covariância:

$$\gamma(0)=\sigma^2(1+\theta^2),\qquad \gamma(\pm1)=\sigma^2\theta,\qquad \gamma(h)=0\ (|h|>1).$$

Substituindo:

$$\boxed{f_{MA(1)}(\lambda)=\frac{\sigma^2}{2\pi}\left(1+\theta^2+2\theta\cos\lambda\right)=\frac{\sigma^2}{2\pi}|1+\theta e^{-i\lambda}|^2.}\tag{9.3.5}$$

Se $\theta=0{,}7$, o espectro tem máximo em $\lambda=0$. Se $\theta=-0{,}7$, seu máximo está em $\lambda=\pi$. Isso conecta o sinal da autocovariância no atraso 1 à localização da potência espectral. Faça o cálculo manual de $f(0)$ e $f(\pi)$ em ambos os casos antes do código.

### 6. Exemplo 3 — AR(1) como soma de cossenos

Considere $X_t=\phi X_{t-1}+\varepsilon_t$ com $|\phi|<1$, $E\varepsilon_t=0$, $\operatorname{Var}(\varepsilon_t)=\sigma^2$ e ruído não correlacionado com o passado. A solução estacionária causal tem

$$\gamma(h)=\frac{\sigma^2}{1-\phi^2}\phi^{|h|}.$$

Substituindo em (9.3.2) e usando $\sum_{h=1}^{\infty}z^h=z/(1-z)$ para $|z|<1$:

$$\begin{aligned}
1+2\sum_{h=1}^{\infty}\phi^h\cos(h\lambda)
&=1+\sum_{h=1}^{\infty}(\phi e^{i\lambda})^h+\sum_{h=1}^{\infty}(\phi e^{-i\lambda})^h\\
&=1+\frac{\phi e^{i\lambda}}{1-\phi e^{i\lambda}}+\frac{\phi e^{-i\lambda}}{1-\phi e^{-i\lambda}}\\
&=\frac{1-\phi^2}{1+\phi^2-2\phi\cos\lambda}.
\end{aligned}$$

Portanto,

$$\boxed{f_{AR(1)}(\lambda)=\frac{\sigma^2}{2\pi(1+\phi^2-2\phi\cos\lambda)}.}\tag{9.3.6}$$

Se $\phi=0{,}8$, a persistência é positiva e o espectro concentra energia perto de zero. Se $\phi=-0{,}8$, há alternância de sinal e a concentração ocorre perto de $\pi$. **Limitação:** a fórmula pressupõe $|\phi|<1$; para $\phi=1$ o passeio aleatório não possui espectro estacionário ordinário.

### 7. Quando não existe densidade usual: distribuição espectral

Morettin e Toloi, §14.3, introduzem uma **função de distribuição espectral** $F$, não decrescente, que permite escrever

$$\boxed{\gamma(h)=\int_{-\pi}^{\pi}e^{ih\lambda}\,dF(\lambda).}\tag{9.3.7}$$

A integral é de **Stieltjes**: ela permite tanto regiões contínuas quanto massas concentradas. Se $F$ é absolutamente contínua, $dF(\lambda)=f(\lambda)\,d\lambda$ e recuperamos (9.3.3). A quantidade total $F(\pi)-F(-\pi^-)$, com convenção de extremos apropriada, é $\gamma(0)$.

**Exemplo construtivo.** Tome $X_t=A\cos(\omega_0t)+B\sin(\omega_0t)$, em que $A,B$ são não correlacionados, $E(A)=E(B)=0$, $\operatorname{Var}(A)=\operatorname{Var}(B)=v$, e $0<\omega_0<\pi$. Então

$$\begin{aligned}
\gamma(h)&=E[X_tX_{t+h}]\\
&=v\cos(\omega_0t)\cos(\omega_0(t+h))+v\sin(\omega_0t)\sin(\omega_0(t+h))\\
&=v\cos(\omega_0h).
\end{aligned}$$

A autocovariância não decai: a soma absoluta diverge. A representação espectral consiste em duas massas $v/2$ em $\lambda=\pm\omega_0$, pois

$$\frac v2e^{ih\omega_0}+\frac v2e^{-ih\omega_0}=v\cos(h\omega_0).$$

Não existe uma função $f$ integrável ordinária representando essas massas como picos de altura finita. Um periodograma amostral pode mostrar picos altos e estreitos, mas não transforma tais massas em densidade contínua. O exemplo reforça a **distinção entre linha espectral e densidade espectral** do livro, pp. 363–366.

### 8. Experimento computacional — ida e volta entre $\gamma$ e $f$

No código seguinte compararemos a fórmula fechada do AR(1) à soma truncada de autocovariâncias e integraremos a densidade para reconstruir $\gamma(h)$. As fórmulas continuam sendo teóricas; a discretização numérica aproxima integrais, e o truncamento dos atrasos aproxima séries infinitas.

```python
import numpy as np
import matplotlib.pyplot as plt

sigma2 = 1.0
phi = 0.8
lam = np.linspace(-np.pi, np.pi, 4097)
f_teorica = sigma2 / (2*np.pi*(1+phi**2-2*phi*np.cos(lam)))

H = 60  # truncamento de |h| <= H
h = np.arange(-H, H+1)
gamma_h = sigma2 / (1-phi**2) * phi**np.abs(h)
f_aprox = np.sum(gamma_h[:,None]*np.cos(h[:,None]*lam[None,:]), axis=0)/(2*np.pi)

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(lam, f_teorica, label='Densidade fechada do AR(1)')
ax.plot(lam, f_aprox, '--', label='Soma de autocovariâncias truncada')
ax.set(xlabel='Frequência angular (rad/amostra)',ylabel='Densidade espectral',title='Do domínio das covariâncias ao espectro')
ax.legend(); ax.grid(alpha=.2); plt.show()

for atraso in [0,1,2,5]:
    gamma_integrada = np.trapezoid(np.cos(atraso*lam)*f_teorica, lam)
    gamma_exata = sigma2/(1-phi**2)*phi**atraso
    print(f'h={atraso}: integral={gamma_integrada:.7f}, fórmula={gamma_exata:.7f}')
    assert np.isclose(gamma_integrada, gamma_exata, atol=1e-5)
assert np.max(np.abs(f_aprox-f_teorica)) < 1e-4
```

**Interpretação:** a superposição das curvas verifica numericamente a transformada direta, enquanto as integrais verificam a transformada inversa. As pequenas discrepâncias dependem da grade numérica e do limite $H$.

```python
omega = np.linspace(0, np.pi, 600)
def f_ma1(theta):
    return (1 + theta**2 + 2*theta*np.cos(omega)) / (2*np.pi)

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(omega, f_ma1(.7), label=r'$\theta=+0,7$')
ax.plot(omega, f_ma1(-.7), label=r'$\theta=-0,7$')
ax.set(xlabel='Frequência angular (rad/amostra)', ylabel='Densidade espectral',title='O sinal da autocovariância altera a distribuição por frequência')
ax.legend(); ax.grid(alpha=.2); plt.show()
```

### 9. Limitações e cuidados de interpretação

- Estacionariedade de segunda ordem é central: espectros estacionários não descrevem diretamente processos como passeios aleatórios em nível.
- Um pico espectral descreve concentração de variância, não comprova, por si só, periodicidade determinística.
- A equivalência entre $\gamma$ e $f$ é uma propriedade **populacional**; estimativas amostrais terão incerteza.
- Reescalar de rad/amostra para ciclos/amostra exige mudar a densidade pelo jacobiano: se $\nu=\lambda/(2\pi)$, então $g(\nu)=2\pi f(2\pi\nu)$.
