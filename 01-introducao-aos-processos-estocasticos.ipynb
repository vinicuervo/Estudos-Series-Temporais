## 1. Motivação: uma sequência observada não descreve toda a incerteza

**Base no livro:** seção 2.2, pp. 25–27; contextualização também presente no início da p. 25.

Imagine que uma loja registrou suas vendas durante cinco dias:

$$
12,\quad 15,\quad 11,\quad 18,\quad 14.
$$

Esses números dizem o que aconteceu. Não dizem, sozinhos, a probabilidade de vender mais de 16 unidades amanhã, nem quanto uma venda elevada hoje altera as possibilidades para o dia seguinte.

Uma fórmula determinística, como $z(t)=10+2t$, atribui um único valor a cada instante. Ela pode representar uma tendência, mas não descreve a incerteza em torno dessa tendência. Já um modelo probabilístico admite diferentes resultados possíveis e atribui probabilidades aos eventos de interesse.

Precisamos descrever duas coisas simultaneamente:

- **A incerteza em cada instante:** quais valores podem ocorrer amanhã e com quais probabilidades?
- **A relação entre instantes:** certos valores hoje tornam alguns resultados amanhã mais ou menos prováveis?

Um processo estocástico oferece uma linguagem para tratar as duas questões. Sua utilidade não depende de supor que os dias sejam independentes. Ao contrário, a dependência temporal é uma das razões para construir esse modelo.

**Complementação didática original.** O exemplo de vendas é apenas uma motivação, sem ajuste de modelo ou conjunto de dados real. Também não se afirma que toda incerteza tenha origem física aleatória: podemos usar um modelo probabilístico para representar informação incompleta sobre um fenômeno.

É útil separar três níveis:

| Nível | Pergunta | Objeto |
|---|---|---|
| Modelo | O que poderia acontecer e com que probabilidades? | Processo estocástico |
| Ocorrência possível | Que sequência surgiria em uma situação possível? | Trajetória |
| Dados disponíveis | Que valores foram efetivamente registrados? | Trecho observado de uma trajetória |

Uma série histórica normalmente fornece apenas uma trajetória, observada em um número finito de instantes. Não fornece acesso direto a todas as possibilidades do modelo.

## 2. Aleatoriedade ao longo do tempo: duas maneiras de olhar o mesmo objeto

**Base no livro:** seção 2.2, pp. 25–27, especialmente as interpretações das Figuras 2.1 e 2.2.

Pense em uma tabela cujas linhas representam situações possíveis e cujas colunas representam instantes:

| Situação possível | Instante 1 | Instante 2 | Instante 3 |
|---|---:|---:|---:|
| $\omega_1$ | $Z(1,\omega_1)$ | $Z(2,\omega_1)$ | $Z(3,\omega_1)$ |
| $\omega_2$ | $Z(1,\omega_2)$ | $Z(2,\omega_2)$ | $Z(3,\omega_2)$ |
| $\omega_3$ | $Z(1,\omega_3)$ | $Z(2,\omega_3)$ | $Z(3,\omega_3)$ |

A tabela é uma representação didática; não exige que existam somente três situações possíveis.

**Fixar a coluna** significa escolher um instante. Ainda não sabemos qual linha ocorrerá; temos uma variável aleatória.

**Fixar a linha** significa escolher uma situação $\omega$. Acompanhamos os valores ao longo do tempo; temos uma trajetória.

**Fixar linha e coluna** produz um único número.

Isso explica por que escrever $Z(t)$ não significa que o tempo precise ser aleatório. Nesta seção, $t$ é um índice escolhido; a aleatoriedade está no resultado $\omega$.

Uma trajetória pode apresentar comportamento muito irregular, mas, depois de fixada como função de $t$, seus valores estão determinados. “Aleatório” não significa necessariamente “visualmente desordenado”: há processos com trajetórias constantes ou retas cujo nível ou inclinação foi escolhido aleatoriamente.

## 3. O espaço de probabilidade, explicado antes da definição

**Base no livro:** a seção 2.2, p. 25, situa as variáveis em um mesmo espaço $(\Omega,\mathcal A,P)$.
**Complementação didática original:** a explicação elementar a seguir.

### 3.1 Resultados, eventos e probabilidades

Um espaço de probabilidade é descrito por:

$$
(\Omega,\mathcal A,P).
$$

- $\Omega$ é o **espaço amostral**: conjunto dos resultados possíveis.
- $\omega\in\Omega$ é **um resultado específico**.
- $\mathcal A$ é a coleção dos **eventos** aos quais atribuímos probabilidades.
- $P$ é a regra que atribui um número entre 0 e 1 a cada evento.

Um evento é um conjunto de resultados. Se $\Omega=\{\omega_-,\omega_+\}$, o evento $\{\omega_+\}$ significa que ocorreu o segundo resultado.

No caso finito, podemos adotar como eventos todos os subconjuntos de $\Omega$. Se os dois resultados forem equiprováveis,

$$
P(\{\omega_-\})=P(\{\omega_+\})=\frac12.
$$

A probabilidade de um evento é a soma das probabilidades dos resultados que o compõem. Por isso,

$$
P(\Omega)=\frac12+\frac12=1,
\qquad
P(\varnothing)=0.
$$

Em espaços gerais, $\mathcal A$ é uma **sigma-álgebra**: contém $\Omega$ e é fechada por complementos e uniões enumeráveis. Isso permite combinar eventos sem sair do domínio de $P$. A probabilidade é aditiva para uniões enumeráveis de eventos mutuamente exclusivos. Não precisaremos aprofundar essa construção para trabalhar com os exemplos finitos.

### 3.2 Uma variável aleatória é uma função

Uma variável aleatória real $X$ associa a cada resultado um número:

$$
X:\Omega\longrightarrow\mathbb R.
$$

Por exemplo,

$$
X(\omega_-)=-1,
\qquad
X(\omega_+)=1.
$$

Antes de conhecer $\omega$, não sabemos qual valor de $X$ será observado. Fixado $\omega_+$, sabemos que $X(\omega_+)=1$.

A condição técnica de **mensurabilidade** exige que perguntas como “$X\le a$” definam eventos:

$$
\{\omega\in\Omega:X(\omega)\le a\}\in\mathcal A,
\qquad a\in\mathbb R.
$$

Essa condição dá sentido a $P(X\le a)$. Nos espaços finitos usados aqui, com todos os subconjuntos disponíveis como eventos, ela é automaticamente satisfeita.

## 4. Definição formal de processo estocástico e leitura da notação

**Base no livro:** Definição 2.1 e texto da seção 2.2, p. 25.

Seja $\mathcal T$ um conjunto de índices. Um **processo estocástico real** é uma família

$$
Z=\{Z(t):t\in\mathcal T\}
$$

de variáveis aleatórias reais definidas em um mesmo espaço de probabilidade $(\Omega,\mathcal A,P)$.

Assim, para cada $t\in\mathcal T$,

$$
Z(t):\Omega\longrightarrow\mathbb R.
$$

A expressão “família” diz que cada variável tem um índice que informa a qual instante ela se refere. Não se trata apenas de guardar números em uma lista: precisamos também de suas leis probabilísticas e de suas relações conjuntas.

### 4.1 O papel de cada símbolo

| Símbolo | Significado |
|---|---|
| $Z$ | Processo completo; a família de variáveis indexadas |
| $\mathcal T$ | Conjunto de instantes ou índices |
| $t,s,t_1,\ldots,t_n$ | Instantes específicos |
| $Z(t)$, também escrito $Z_t$ | Variável aleatória associada ao instante $t$ |
| $\Omega$ | Espaço dos resultados possíveis |
| $\omega$ | Um resultado do espaço amostral |
| $Z(t,\omega)$ | Valor do processo no instante $t$, na situação $\omega$ |
| $z(t)$ | Notação adotada aqui para uma trajetória já fixada |
| $z_1,\ldots,z_n$ | Limiares reais usados em eventos; não são os índices de tempo |
| $\mathcal E$ | Espaço de estados: conjunto no qual os valores do processo vivem |

A notação de dois argumentos torna explícita a identidade

$$
Z(t,\omega)=[Z(t)](\omega).
$$

Primeiro escolhemos a variável correspondente a $t$; depois a avaliamos no resultado $\omega$.

O livro também representa realizações por $Z^{(j)}(t)$, em que $j$ identifica uma trajetória. O sobrescrito entre parênteses não é potência. Aqui usaremos principalmente $z(t)$ ou $z_\omega(t)$ para destacar quando a trajetória já está fixada.

### 4.2 Por que o espaço de probabilidade deve ser comum?

Se $Z(1)$ e $Z(2)$ são funções sobre o mesmo $\Omega$, faz sentido perguntar:

$$
P\bigl(Z(1)>10,\ Z(2)>10\bigr).
$$

O mesmo resultado $\omega$ determina os dois valores. Assim podemos estudar a ocorrência conjunta, e não apenas cada dia isoladamente.

Dizer somente “em cada dia, os valores têm determinada distribuição” ainda não especifica como os dias se relacionam. O Exemplo 2 mostrará isso por cálculo.

### 4.3 Tempo discreto não significa valores discretos

**Base no livro:** pp. 26–27.

Há dois eixos de classificação:

| Índice temporal | Valores possíveis | Exemplo de interpretação |
|---|---|---|
| Discreto | Discretos | Número de chamadas registrado por hora |
| Discreto | Contínuos | Temperatura medida a cada hora |
| Contínuo | Discretos | Contagem acumulada de chegadas, acompanhada continuamente |
| Contínuo | Contínuos | Temperatura idealizada como função do tempo contínuo |

Os dois últimos exemplos são complementações didáticas.

Um conjunto de índices pode ser finito, como $\{1,\ldots,N\}$, ou infinito enumerável, como os inteiros $\mathbb Z=\{\ldots,-1,0,1,\ldots\}$. Em tempo contínuo, podemos tomar um intervalo real.

O espaço de estados $\mathcal E$ não deve ser confundido com $\mathcal T$ nem com $\Omega$: instante, resultado possível e valor observado desempenham papéis diferentes.

## 5. Processo, variável, trajetória e observação

**Base no livro:** seção 2.2, pp. 25–27.
**Complementação didática original:** organização das distinções.

Fixado um resultado $\omega_0$, definimos

$$
z_{\omega_0}:\mathcal T\longrightarrow\mathbb R,
\qquad
z_{\omega_0}(t)=Z(t,\omega_0).
$$

Essa função é uma **realização** ou **trajetória**. Nesta seção, os termos são sinônimos.

| Objeto | O que está fixado? | O que ainda varia? |
|---|---|---|
| Processo $Z$ | A regra probabilística | Instante e resultado |
| Variável $Z(t_0)$ | O instante $t_0$ | O resultado $\omega$ |
| Trajetória $z_{\omega_0}(t)$ | O resultado $\omega_0$ | O instante $t$ |
| Número $Z(t_0,\omega_0)$ | Instante e resultado | Nada |

O conjunto das trajetórias é chamado de **ensemble** no livro. É uma coleção de possibilidades do modelo; não significa que possuímos medições de todas elas.

Na prática, observamos algo como

$$
\bigl(z_{\omega_0}(t_1),\ldots,z_{\omega_0}(t_N)\bigr).
$$

Esse vetor é um trecho de uma trajetória. Seu comprimento $N$ não é o número de trajetórias.

Também é importante separar:

- **Média entre possibilidades em um instante:** uma esperança calculada com $P$.
- **Média de observações ao longo de uma trajetória:** uma média temporal.

Essas médias não são automaticamente iguais. Veremos uma diferença explícita no Exemplo 1; não pressuporemos qualquer resultado de ergodicidade.

## 6. Distribuições finito-dimensionais: como descrever instantes em conjunto

**Base no livro:** seção 2.3, pp. 27–28, equação (2.1).

### 6.1 Da distribuição de uma variável à distribuição de um vetor

Para uma variável real $X$, sua função de distribuição é

$$
F_X(x)=P(X\le x).
$$

Ela responde à pergunta: qual é a probabilidade de o valor não ultrapassar $x$?

Escolha agora $n$ instantes $t_1,\ldots,t_n$. Eles formam o vetor aleatório

$$
\bigl(Z(t_1),\ldots,Z(t_n)\bigr).
$$

Sua função de distribuição conjunta é

$$
F(z_1,\ldots,z_n;t_1,\ldots,t_n)
=
P\bigl(Z(t_1)\le z_1,\ldots,Z(t_n)\le z_n\bigr).
$$

Essa é a equação (2.1) do livro, conferida visualmente na p. 27. O ponto e vírgula separa os limiares dos instantes.

As vírgulas dentro de $P$ significam **“e”**: todas as condições devem ocorrer simultaneamente. O índice $n$ conta quantos instantes estamos examinando; não precisa ser o tamanho da série observada.

- Para $n=1$, descrevemos uma distribuição unidimensional ou marginal.
- Para $n=2$, descrevemos a distribuição de um par de observações.
- Para $n=3$, descrevemos um trio.
- O conjunto de todas essas distribuições, para todas as escolhas finitas de instantes, descreve a lei finito-dimensional do processo.

“Finito-dimensional” não significa que o processo só tem um número finito de instantes. Significa que cada consulta envolve um vetor com um número finito de componentes.

### 6.2 Por que as marginais não bastam?

As marginais respondem a perguntas individuais. Duas variáveis podem ter exatamente as mesmas distribuições individuais e, ainda assim, ocorrer sempre juntas ou variar independentemente.

Para previsão, isso é decisivo: interessa saber o que uma observação informa sobre outra. No Exemplo 2 construiremos dois processos com marginais iguais e probabilidades conjuntas diferentes.

Não presumiremos independência. Quando ela for uma hipótese de um exemplo, será declarada ou demonstrada a partir de seu espaço amostral.

### 6.3 Distribuição não é sinônimo de densidade

**Base no livro:** a seção 2.3, p. 28, utiliza uma densidade nos cálculos integrais sob a suposição de que ela existe.
**Complementação didática original:** distinção necessária para os exemplos discretos.

A função de distribuição existe para toda variável aleatória real. Já uma densidade não existe em todos os casos.

Nos exemplos finitos, calcularemos probabilidades somando as massas dos resultados. Não usaremos uma integral nem exigiremos uma densidade.

Quando há densidade $f_X$,

$$
P(a<X\le b)=\int_a^b f_X(x)\,dx.
$$

A integral representa a área sob a densidade no intervalo; substitui a soma de massas no caso contínuo apropriado. Uma probabilidade não é, em geral, o valor da densidade em um ponto.

## 7. Coerência das distribuições: simetria e compatibilidade

**Base no livro:** seção 2.3, pp. 27–28, equações (2.2) e (2.3).
**Complementação didática original:** demonstrações elementares e esclarecimentos.

Não podemos escolher arbitrariamente uma distribuição para cada par ou trio. As respostas precisam concordar quando representam a mesma pergunta.

### 7.1 Simetria: reordenar os pares instante–limiar não muda o evento

Para dois instantes,

$$
F(a,b;s,t)
=
P(Z(s)\le a,\ Z(t)\le b).
$$

Trocar a ordem das duas condições não altera a interseção:

$$
P(Z(s)\le a,\ Z(t)\le b)
=
P(Z(t)\le b,\ Z(s)\le a).
$$

Logo,

$$
F(a,b;s,t)=F(b,a;t,s).
$$

Para uma permutação $j_1,\ldots,j_n$ dos índices $1,\ldots,n$,

$$
F(z_{j_1},\ldots,z_{j_n};t_{j_1},\ldots,t_{j_n})
=
F(z_1,\ldots,z_n;t_1,\ldots,t_n).
$$

Permutação significa apenas uma reordenação dos índices. É indispensável reordenar **limiares e instantes juntos**.

Essa propriedade não afirma que $F(a,b;s,t)=F(b,a;s,t)$. Esta última igualdade, quando verdadeira, expressa uma simetria adicional da distribuição e não decorre apenas da ordem das condições.

### 7.2 Compatibilidade: retirar uma restrição recupera uma marginal

Considere

$$
F(a,b;s,t)=P(Z(s)\le a,\ Z(t)\le b).
$$

À medida que $b$ cresce sem limite, a condição $Z(t)\le b$ deixa de excluir resultados, pois $Z(t)$ é real e finito. Assim,

$$
\lim_{b\to+\infty}F(a,b;s,t)
=
P(Z(s)\le a)
=
F(a;s).
$$

A justificativa probabilística é que os eventos aumentam até o evento $Z(s)\le a$; a probabilidade da união crescente é o limite de suas probabilidades.

Para $m<n$, a equação (2.3) generaliza essa operação:

$$
\lim_{z_{m+1},\ldots,z_n\to+\infty}
F(z_1,\ldots,z_n;t_1,\ldots,t_n)
=
F(z_1,\ldots,z_m;t_1,\ldots,t_m).
$$

Escrever $+\infty$ como argumento da distribuição é uma abreviação para esse limite, não um novo estado real do processo.

### 7.3 Um cuidado quando um instante se repete

Se duas componentes correspondem ao mesmo instante,

$$
\{Z(t)\le a,\ Z(t)\le b\}
=
\{Z(t)\le\min(a,b)\}.
$$

Portanto,

$$
F(a,b;t,t)=F(\min(a,b);t).
$$

Não podemos tratar duas cópias da mesma variável como duas medições independentes. Essa identidade é uma consequência didática da definição e ajuda a interpretar a consistência em índices repetidos.

O livro registra, na p. 28, um resultado de existência associado às distribuições consistentes. Não o demonstramos aqui. Sua formulação geral exige que sejam distribuições válidas, com a coerência apropriada entre todas as seleções de índices; para seleções de tempos distintos, simetria e marginalização expressam as condições usuais. Não afirmamos que tais condições, sozinhas, garantam trajetórias contínuas ou diferenciáveis.

## 8. Momentos: descrições mais simples, com hipóteses explícitas

**Base no livro:** seção 2.3, pp. 28–29, equações (2.4)–(2.8).
**Complementação didática original:** detalhamento das operações.

Conhecer todas as distribuições conjuntas é exigente. Momentos oferecem descrições parciais que frequentemente são mais fáceis de calcular.

### 8.1 Esperança e função média

Em um espaço finito,

$$
E[X]=\sum_{\omega\in\Omega}X(\omega)P(\{\omega\}).
$$

O símbolo $\sum$ manda somar um termo para cada resultado. Multiplicamos o valor pela sua probabilidade porque resultados mais prováveis têm maior peso.

Para o processo,

$$
\mu(t)=E[Z(t)].
$$

Essa é uma função do tempo, não necessariamente uma constante. É a ideia da equação (2.5). No caso contínuo com densidade, a soma é substituída por uma integral ponderada.

Para que a média seja finita de maneira bem definida, adotaremos $E[|Z(t)|]<\infty$. Nos espaços finitos com valores finitos isso é automático.

### 8.2 Variância

A variância mede dispersão ao redor da média:

$$
V(t)=E\bigl[(Z(t)-\mu(t))^2\bigr].
$$

O quadrado impede que desvios positivos e negativos se cancelem. Se $E[Z(t)^2]<\infty$, podemos expandir:

$$
\begin{aligned}
V(t)
&=E[Z(t)^2-2\mu(t)Z(t)+\mu(t)^2]\\
&=E[Z(t)^2]-2\mu(t)E[Z(t)]+\mu(t)^2\\
&=E[Z(t)^2]-2\mu(t)^2+\mu(t)^2\\
&=E[Z(t)^2]-\mu(t)^2.
\end{aligned}
$$

Usamos a linearidade da esperança e o fato de $\mu(t)$ ser um número determinado pelo modelo. A identidade corresponde à equação (2.7).

### 8.3 Autocovariância

Para dois instantes com segundos momentos finitos,

$$
\gamma(s,t)=\operatorname{Cov}(Z(s),Z(t))
=
E[(Z(s)-\mu(s))(Z(t)-\mu(t))].
$$

Expandindo o produto,

$$
\begin{aligned}
\gamma(s,t)
&=E[Z(s)Z(t)]-\mu(t)E[Z(s)]-\mu(s)E[Z(t)]
  +\mu(s)\mu(t)\\
&=E[Z(s)Z(t)]-\mu(s)\mu(t).
\end{aligned}
$$

Essa é a identidade da equação (2.6). Momentos de segunda ordem finitos também garantem a integrabilidade do produto, pois $2|xy|\le x^2+y^2$.

A palavra “auto” indica que comparamos o processo com ele mesmo em dois instantes. Por enquanto, conservamos os dois argumentos $s,t$; não reduzimos a função a uma defasagem.

Se $s=t$,

$$
\gamma(t,t)=V(t).
$$

Covariância positiva indica associação linear entre desvios de mesmo sinal; covariância negativa indica associação linear entre desvios de sinais opostos. Covariância zero não exclui toda forma de dependência.

### 8.4 Momentos produtos e limites da descrição por momentos

O livro introduz, na equação (2.4),

$$
E\bigl[Z(t_1)^{r_1}\cdots Z(t_n)^{r_n}\bigr],
$$

sob condições de existência. Para expoentes inteiros não negativos, essa expressão é um momento produto: calculamos potências, multiplicamos e tiramos a esperança.

Por exemplo, $E[Z(s)Z(t)]$ ajuda a calcular a autocovariância. O vetor de expoentes especifica qual aspecto estamos resumindo.

Uma lista de médias e covariâncias não descreve, em geral, toda a lei do processo. Mesmo conhecer muitos momentos não autoriza supor automaticamente que toda a distribuição foi determinada. Momentos são resumos com alcance condicionado às hipóteses adotadas.

## 9. Exemplo 1 — Uma inclinação aleatória, calculada por inteiro

**Complementação didática original.** Este exemplo não é atribuído ao livro.

### 9.1 Construção

Tome

$$
\Omega=\{-1,+1\},
\qquad
P(\{-1\})=P(\{+1\})=\frac12,
\qquad
\mathcal T=\{0,1,2\}.
$$

Defina

$$
Z(t,\omega)=t\omega.
$$

Para cada $t$, a função de $\omega$ é uma variável aleatória no mesmo espaço. Logo, a construção satisfaz a definição de processo.

| Resultado | Probabilidade | $Z(0)$ | $Z(1)$ | $Z(2)$ |
|---|---:|---:|---:|---:|
| $-1$ | $1/2$ | 0 | -1 | -2 |
| $+1$ | $1/2$ | 0 | 1 | 2 |

Há duas trajetórias: $z_-(t)=-t$ e $z_+(t)=t$. Ambas são retas nos instantes considerados, mas não sabemos antecipadamente qual ocorrerá.

### 9.2 Distribuições individuais e conjunta

No instante 1,

$$
P(Z(1)=-1)=P(Z(1)=1)=\frac12.
$$

No instante 2,

$$
P(Z(2)=-2)=P(Z(2)=2)=\frac12.
$$

Para calcular $F(0,1;1,2)$, precisamos atender às duas restrições:

$$
Z(1)\le0
\quad\text{e}\quad
Z(2)\le1.
$$

A linha $\omega=-1$ atende a ambas, pois $-1\le0$ e $-2\le1$. A linha $\omega=+1$ não atende. Portanto,

$$
F(0,1;1,2)=\frac12.
$$

Também,

$$
P(Z(1)=1,Z(2)=2)=\frac12.
$$

Se multiplicássemos as marginais, obteríamos $1/4$. Isso não é a probabilidade conjunta: os valores estão ligados pela mesma inclinação aleatória.

### 9.3 Média, variância e autocovariância

Para qualquer $t\in\{0,1,2\}$,

$$
\mu(t)=(-t)\frac12+t\frac12=0.
$$

O segundo momento é

$$
E[Z(t)^2]=t^2\frac12+t^2\frac12=t^2.
$$

Como a média é zero,

$$
V(t)=t^2.
$$

Para $s,t\in\mathcal T$,

$$
Z(s,\omega)Z(t,\omega)=s\omega\,t\omega=st\omega^2=st,
$$

porque $\omega^2=1$ nos dois resultados. Logo,

$$
E[Z(s)Z(t)]=st,
\qquad
\gamma(s,t)=st-0\cdot0=st.
$$

Conferimos a diagonal: $\gamma(t,t)=t^2=V(t)$.

### 9.4 Média temporal não é automaticamente esperança

Na trajetória positiva, a média dos três valores é

$$
\frac{0+1+2}{3}=1.
$$

Na negativa, ela é $-1$. Entretanto, $\mu(t)=0$ em cada instante.

Não há contradição: a média temporal usa diferentes instantes de uma linha; a esperança usa os resultados possíveis de uma coluna, ponderados por suas probabilidades.

## 10. Exemplo 2 — Mesmas marginais, dependências diferentes

**Complementação didática original.** Construiremos dois processos em um mesmo espaço para mostrar exatamente o que as marginais deixam de informar.

### 10.1 Espaço amostral e processos

Considere quatro resultados equiprováveis,

$$
\Omega=\{(-1,-1),(-1,1),(1,-1),(1,1)\}.
$$

Cada resultado é um par $(u,v)$ e tem probabilidade $1/4$. Para $\mathcal T=\{1,2\}$, defina:

$$
A(1,(u,v))=u,\qquad A(2,(u,v))=u,
$$

$$
B(1,(u,v))=u,\qquad B(2,(u,v))=v.
$$

O processo $A$ repete a primeira coordenada; $B$ usa as duas coordenadas.

| $(u,v)$ | Probabilidade | $A(1)$ | $A(2)$ | $B(1)$ | $B(2)$ |
|---|---:|---:|---:|---:|---:|
| $(-1,-1)$ | $1/4$ | -1 | -1 | -1 | -1 |
| $(-1,1)$ | $1/4$ | -1 | -1 | -1 | 1 |
| $(1,-1)$ | $1/4$ | 1 | 1 | 1 | -1 |
| $(1,1)$ | $1/4$ | 1 | 1 | 1 | 1 |

Dois resultados distintos podem produzir a mesma trajetória de $A$. Portanto, não precisamos de uma correspondência um a um entre resultados do espaço amostral e trajetórias.

### 10.2 Verificação das marginais

Em cada coluna, dois resultados produzem $-1$ e dois produzem $1$. Assim, para $t=1,2$,

$$
P(A(t)=-1)=P(B(t)=-1)=\frac12,
$$

e o mesmo vale para o valor $1$. As quatro variáveis têm média zero e variância 1.

### 10.3 Verificação da distribuição conjunta

Para $A$, o evento $A(1)\le0,A(2)\le0$ contém as duas primeiras linhas:

$$
F_A(0,0;1,2)=\frac14+\frac14=\frac12.
$$

Para $B$, somente a primeira linha atende às duas restrições:

$$
F_B(0,0;1,2)=\frac14.
$$

Portanto, as marginais são iguais, mas as distribuições bidimensionais são diferentes. Isso prova, neste exemplo, que as marginais não determinam a lei conjunta.

Em $B$, cada combinação de sinais tem probabilidade $1/4=(1/2)(1/2)$; as duas coordenadas são independentes. Em $A$, a segunda variável é igual à primeira.

### 10.4 O que observar o primeiro instante informa?

Em um espaço finito, para $P(D)>0$,

$$
P(C\mid D)=\frac{P(C\cap D)}{P(D)}.
$$

Condicionar significa restringir a atenção aos resultados em que $D$ ocorreu e recalcular suas proporções.

Para $A$,

$$
P(A(2)=1\mid A(1)=1)
=\frac{1/2}{1/2}=1.
$$

Para $B$,

$$
P(B(2)=1\mid B(1)=1)
=\frac{1/4}{1/2}=\frac12.
$$

A mesma distribuição individual esconde possibilidades preditivas diferentes.

### 10.5 Covariâncias e compatibilidade

Para $A$, o produto $A(1)A(2)$ vale 1 em todas as linhas. Portanto,

$$
\operatorname{Cov}(A(1),A(2))=1-0\cdot0=1.
$$

Para $B$, os produtos são $1,-1,-1,1$, com pesos iguais:

$$
E[B(1)B(2)]=\frac{1-1-1+1}{4}=0.
$$

Logo, sua covariância é zero. A independência de $B$ foi demonstrada pela distribuição conjunta; não foi deduzida apenas da covariância.

Para recuperar a marginal do primeiro instante, permitimos qualquer valor da segunda variável. Como os estados são $-1$ e 1, já basta usar o limiar 1:

$$
F_A(0,1;1,2)=\frac12,
\qquad
F_B(0,1;1,2)=\frac12.
$$

São as respectivas probabilidades de a primeira variável não ultrapassar zero, como exige a compatibilidade.

## 11. Conferência computacional opcional dos exemplos

**Complementação didática original.** O código enumera todos os resultados; não realiza simulação aleatória. Usa apenas a biblioteca padrão do Python.

O objetivo é confrontar as contas manuais com uma enumeração exata. Frações evitam erros de arredondamento. Essa conferência não demonstra os resultados gerais sobre processos.

```python
from fractions import Fraction
from itertools import product

# Exemplo 1: os dois resultados têm a mesma probabilidade.
omega_z = (-1, 1)
peso_z = Fraction(1, 2)

for t in (0, 1, 2):
    media = sum((t * w) * peso_z for w in omega_z)
    segundo_momento = sum((t * w) ** 2 * peso_z for w in omega_z)
    variancia = segundo_momento - media ** 2
    print(f"Z, t={t}: media={media}, variancia={variancia}")

# Exemplo 2: quatro resultados equiprováveis.
omega_ab = tuple(product((-1, 1), repeat=2))
peso_ab = Fraction(1, 4)

for nome in ("A", "B"):
    pares = [(u, u) if nome == "A" else (u, v)
             for u, v in omega_ab]
    media_1 = sum(x * peso_ab for x, y in pares)
    media_2 = sum(y * peso_ab for x, y in pares)
    produto_medio = sum(x * y * peso_ab for x, y in pares)
    covariancia = produto_medio - media_1 * media_2
    conjunta = sum((peso_ab for x, y in pares if x <= 0 and y <= 0),
                   Fraction(0))
    marginal = sum((peso_ab for x, y in pares if x <= 0),
                   Fraction(0))
    print(f"{nome}: F(0,0)={conjunta}, marginal={marginal}, "
          f"covariancia={covariancia}")
```


Saída esperada, conferida por execução:

~~~text
Z, t=0: media=0, variancia=0
Z, t=1: media=0, variancia=1
Z, t=2: media=0, variancia=4
A: F(0,0)=1/2, marginal=1/2, covariancia=1
B: F(0,0)=1/4, marginal=1/2, covariancia=0
~~~

No código, cada linha do espaço amostral continua contando separadamente, mesmo quando duas linhas produzem a mesma trajetória. Ignorar essa multiplicidade mudaria as probabilidades.
