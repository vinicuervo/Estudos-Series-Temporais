# Módulo 01 — Seção 2: Estacionariedade estrita e estacionariedade fraca

**Referência principal:** Morettin e Toloi, *Análise de Séries Temporais — Modelos lineares univariados*, volume 1, 3ª edição, Definições 2.2 e 2.3, página impressa 30. A observação da página 31 esclarece a relação entre elas e o caso gaussiano.

## 1. Intuição: deslocar o relógio sem alterar a descrição probabilística

Considere dois trechos de um processo:

$$
(Z_1,Z_2,Z_3)
\qquad\text{e}\qquad
(Z_{11},Z_{12},Z_{13}).
$$

Os valores observados nesses trechos não precisam ser iguais. A pergunta é se as **probabilidades dos possíveis valores conjuntos** são as mesmas. Nos dois trechos, os instantes mantêm as mesmas distâncias: apenas a origem do relógio mudou.

A estacionariedade estrita exige que essa invariância alcance toda distribuição conjunta de um número finito de instantes. A estacionariedade fraca exige menos: invariância da média e da estrutura de covariâncias, com momentos de segunda ordem finitos.

**Estacionariedade é uma propriedade do modelo probabilístico, não a afirmação de que a trajetória é constante.** Um gráfico oscilando ao redor de um nível pode sugerir uma hipótese; não demonstra nenhuma das definições.

Para simplificar as demonstrações, trabalharemos com processos reais $Z=\{Z_t:t\in\mathbb Z\}$. O índice $t$ representa o tempo, e $\mathbb Z$ contém todos os inteiros. Em outros conjuntos de índices, consideram-se somente os deslocamentos que mantêm os instantes envolvidos no domínio.

## 2. Estacionariedade estrita: invariância de todas as distribuições finito-dimensionais

**Base no livro:** Definição 2.2, p. 30, equação (2.9).

O processo é **estritamente estacionário**, ou estacionário em sentido forte, quando, para todo inteiro $n\ge1$, quaisquer instantes $t_1,\ldots,t_n$ e qualquer deslocamento inteiro $\tau$,

$$
(Z_{t_1+\tau},\ldots,Z_{t_n+\tau})
\overset{d}{=}
(Z_{t_1},\ldots,Z_{t_n}).
$$

O símbolo $\overset{d}{=}$ significa **igualdade de distribuição**, não igualdade dos valores das variáveis. Na notação de função de distribuição usada pelo livro:

$$
F(z_1,\ldots,z_n;t_1+\tau,\ldots,t_n+\tau)
=
F(z_1,\ldots,z_n;t_1,\ldots,t_n),
$$

para todos os limiares reais $z_1,\ldots,z_n$, em que

$$
F(z_1,\ldots,z_n;t_1,\ldots,t_n)
=
P(Z_{t_1}\le z_1,\ldots,Z_{t_n}\le z_n).
$$

A primeira igualdade é uma reexpressão da definição; a segunda relembra o significado de $F$.

### O que cada exigência acrescenta?

- **$n=1$:** a distribuição individual de $Z_t$ não depende de $t$.
- **$n=2$:** também deve permanecer igual a distribuição conjunta de cada par ao deslocar os dois instantes pelo mesmo $\tau$.
- **$n=3$ e seguintes:** a exigência inclui relações que só aparecem em trios ou conjuntos maiores.

Não basta verificar $n=1$. Nem basta verificar apenas os pares para concluir a condição de todas as ordens.

O deslocamento é **comum a todos os instantes**. Mover $(1,3)$ para $(11,13)$ preserva a distância temporal; mover para $(11,20)$ não é o mesmo tipo de comparação.

A definição não exige média, variância, densidade, normalidade ou independência. Ela fala diretamente de distribuições, que podem existir mesmo quando os momentos não são finitos.

## 3. Estacionariedade fraca: invariância dos momentos de primeira e segunda ordens

**Base no livro:** Definição 2.3, p. 30, equação (2.12).

Um processo real é **fracamente estacionário**, estacionário de segunda ordem ou estacionário em sentido amplo quando satisfaz simultaneamente:

$$
\boxed{
\begin{aligned}
\text{(i)}\quad &E[Z_t]=\mu,\qquad \text{para todo }t;\\
\text{(ii)}\quad &E[Z_t^2]<\infty,\qquad \text{para todo }t;\\
\text{(iii)}\quad
&\operatorname{Cov}(Z_s,Z_t)=c(t-s),\qquad \text{para todos }s,t.
\end{aligned}
}
$$

Aqui, $\mu$ é uma constante real e $c$ é uma função da **defasagem** $h=t-s$, a distância orientada entre os instantes.

### 3.1 Por que precisamos das três condições?

**Média constante.** O centro probabilístico não se desloca com o calendário. Isso não impede oscilações de uma trajetória ao redor desse centro.

**Segundo momento finito.** Garante variância finita e permite trabalhar com covariâncias. Também garante que a média existe e é finita, pela desigualdade

$$
E[|Z_t|]\le\sqrt{E[Z_t^2]}<\infty.
$$

Essa é uma aplicação da desigualdade de Cauchy–Schwarz à variável $|Z_t|$ e à constante 1.

**Covariância dependente apenas da defasagem.** Dois pares com o mesmo afastamento temporal têm a mesma covariância. Por exemplo,

$$
\operatorname{Cov}(Z_2,Z_5)
=
c(3)
=
\operatorname{Cov}(Z_{20},Z_{23}).
$$

Não exigimos que essa covariância seja zero: estacionariedade fraca permite dependência temporal.

### 3.2 A variância constante já está incluída

Tomando $s=t$ na terceira condição,

$$
\operatorname{Var}(Z_t)
=
\operatorname{Cov}(Z_t,Z_t)
=
c(0).
$$

Portanto, a variância é constante e finita. Além disso,

$$
E[Z_t^2]=c(0)+\mu^2
$$

é constante.

Entretanto, **média e variância constantes não bastam**. Ainda é necessário verificar as covariâncias entre instantes diferentes; o Exemplo 4 mostra exatamente essa falha.

### 3.3 Por que o livro escreve uma função de $|t-s|$?

Para variáveis reais, a covariância é simétrica:

$$
\operatorname{Cov}(Z_s,Z_t)
=
\operatorname{Cov}(Z_t,Z_s).
$$

Se ela depende somente de $h=t-s$, segue que

$$
c(h)=c(-h).
$$

Assim, podemos escrever uma função da defasagem assinada, com simetria, ou uma função da distância $|t-s|$, como na Definição 2.3. As formulações são equivalentes neste contexto.

Essa simetria da covariância não afirma que toda distribuição conjunta seja invariável por inversão da ordem temporal.

## 4. Relação matemática entre as definições

### 4.1 Estrita com segundo momento finito implica fraca

**Base no livro:** discussão da p. 30, equações (2.10)–(2.11), e ressalva no início da p. 31.
**Complementação didática:** demonstração detalhada.

Suponha que $Z$ seja estritamente estacionário e que

$$
E[Z_0^2]<\infty.
$$

É suficiente impor a finitude em um instante porque as distribuições individuais são iguais.

**Passo 1 — Igualdade das marginais.** A definição com $n=1$ fornece

$$
Z_t\overset{d}{=}Z_0.
$$

Logo,

$$
E[Z_t^2]=E[Z_0^2]<\infty
$$

e

$$
E[Z_t]=E[Z_0]=\mu.
$$

A existência da média decorre do segundo momento finito. Temos as condições (i) e (ii).

**Passo 2 — Igualdade da distribuição dos pares.** Na definição com $n=2$, desloque o par $(s,t)$ por $\tau=-s$:

$$
(Z_s,Z_t)\overset{d}{=}(Z_0,Z_{t-s}).
$$

**Passo 3 — Igualdade das covariâncias.** Os produtos são integráveis porque

$$
2|Z_sZ_t|\le Z_s^2+Z_t^2.
$$

Portanto, a igualdade de distribuição permite calcular as mesmas esperanças de produtos:

$$
\begin{aligned}
\operatorname{Cov}(Z_s,Z_t)
&=E[Z_sZ_t]-\mu^2\\
&=E[Z_0Z_{t-s}]-\mu^2\\
&=\operatorname{Cov}(Z_0,Z_{t-s}).
\end{aligned}
$$

Defina $c(h)=\operatorname{Cov}(Z_0,Z_h)$. Então,

$$
\operatorname{Cov}(Z_s,Z_t)=c(t-s),
$$

que é a terceira condição. Concluímos:

$$
\boxed{
\text{estacionariedade estrita}
+
E[Z_0^2]<\infty
\ \Longrightarrow\
\text{estacionariedade fraca}.
}
$$

Sem a hipótese de segundo momento finito, a conclusão não é válida.

### 4.2 A recíproca não vale em geral

A estacionariedade fraca preserva média e covariância. Distribuições diferentes podem ter os mesmos dois primeiros momentos. Portanto, esses momentos não garantem a igualdade de todas as distribuições finito-dimensionais.

O Exemplo 3 fornece uma construção explícita: a distribuição individual muda com a paridade do instante, mas média e estrutura de covariância permanecem invariantes.

### 4.3 O caso gaussiano

**Base no livro:** p. 31, Definição 2.4 e parágrafo seguinte.

Um processo é **gaussiano** quando todo vetor finito $(Z_{t_1},\ldots,Z_{t_n})$ tem distribuição normal multivariada. Não basta que cada variável isolada tenha distribuição normal.

Uma lei normal multivariada é determinada pelo vetor de médias e pela matriz de covariâncias. Sob essa propriedade, podemos provar a recíproca.

Se o processo gaussiano é fracamente estacionário, o vetor

$$
(Z_{t_1+\tau},\ldots,Z_{t_n+\tau})
$$

tem o mesmo vetor de médias $(\mu,\ldots,\mu)$ que o vetor não deslocado. Para cada par de componentes $i,j$, sua covariância é

$$
\begin{aligned}
\operatorname{Cov}(Z_{t_i+\tau},Z_{t_j+\tau})
&=c\bigl((t_j+\tau)-(t_i+\tau)\bigr)\\
&=c(t_j-t_i).
\end{aligned}
$$

As matrizes de covariância também são iguais. Como os dois vetores são gaussianos, possuem a mesma distribuição. Isso vale para qualquer número de instantes e qualquer deslocamento:

$$
\boxed{
\text{processo gaussiano e fracamente estacionário}
\ \Longrightarrow\
\text{estritamente estacionário}.
}
$$

A propriedade de determinação da lei normal por média e covariância é utilizada aqui como resultado; não estamos demonstrando a teoria da normal multivariada. Para processos gaussianos, os segundos momentos são finitos, de modo que as duas formas de estacionariedade são equivalentes.

## 5. Exemplos e contraexemplos

**Complementações didáticas originais.** As construções abaixo ilustram as definições e relações do livro; não são atribuídas às páginas consultadas.

### Exemplo 1 — Um processo que satisfaz as duas definições

Sejam $\{\varepsilon_t:t\in\mathbb Z\}$ variáveis independentes, cada uma com

$$
P(\varepsilon_t=-1)=P(\varepsilon_t=1)=\frac12.
$$

Defina $Z_t=\varepsilon_t$.

**Estacionariedade estrita.** Para instantes distintos, as probabilidades conjuntas são produtos das mesmas probabilidades marginais:

$$
P(Z_{t_1}\le z_1,\ldots,Z_{t_n}\le z_n)
=
\prod_{j=1}^{n}P(\varepsilon_0\le z_j).
$$

O produto não depende das posições dos instantes. Um deslocamento comum preserva essa expressão. Se algum instante se repete, o evento pode ser reduzido usando o menor dos limiares associados àquele instante; a repetição é preservada pelo deslocamento. Logo, a definição vale para todas as escolhas.

**Estacionariedade fraca.** Calcule:

$$
E[Z_t]=(-1)\frac12+1\frac12=0,
$$

$$
E[Z_t^2]=1\frac12+1\frac12=1,
\qquad
\operatorname{Var}(Z_t)=1.
$$

Se $s\ne t$, a independência fornece

$$
E[Z_sZ_t]=E[Z_s]E[Z_t]=0,
$$

portanto a covariância é zero. Na diagonal ela vale 1:

$$
c(h)=
\begin{cases}
1,&h=0,\\
0,&h\ne0.
\end{cases}
$$

As três condições estão satisfeitas.

A independência facilita esta construção, mas não faz parte da definição de estacionariedade. Por exemplo, se $U$ assume $-1$ e $1$ com probabilidades iguais e $Y_t=U$ para todo $t$, o vetor em qualquer conjunto de instantes é sempre $(U,\ldots,U)$. Esse processo também é estritamente e fracamente estacionário, com $c(h)=1$ para todo $h$, apesar da dependência perfeita.

### Exemplo 2 — Estritamente estacionário, mas sem estacionariedade fraca

Seja $U$ uma variável com densidade de Cauchy padrão,

$$
f(u)=\frac{1}{\pi(1+u^2)},
\qquad u\in\mathbb R,
$$

e defina $Z_t=U$ para todo $t\in\mathbb Z$.

A densidade é uma função não negativa cuja integral é 1. Neste caso, isso pode ser conferido pela primitiva $\arctan(u)/\pi$, cujos limites são $-1/2$ e $1/2$.

**Passo 1 — Verificar a estacionariedade estrita.**

Qualquer vetor de observações é

$$
(Z_{t_1},\ldots,Z_{t_n})=(U,\ldots,U).
$$

Depois do deslocamento, o vetor é o mesmo, inclusive como variável aleatória. Logo, suas distribuições são iguais para todas as ordens.

**Passo 2 — Examinar o segundo momento.**

Por definição de esperança para uma densidade,

$$
E[U^2]
=
\int_{-\infty}^{\infty}\frac{u^2}{\pi(1+u^2)}\,du.
$$

Para $u\ge1$, temos $1+u^2\le2u^2$, portanto

$$
\frac{u^2}{\pi(1+u^2)}\ge\frac{1}{2\pi}.
$$

Consequentemente,

$$
E[U^2]
\ge
\int_1^\infty\frac{1}{2\pi}\,du
=
\infty.
$$

A integral de uma constante positiva sobre um intervalo ilimitado diverge. Falha a condição (ii) da estacionariedade fraca.

**Não se deve dizer que a média de Cauchy é zero apenas por simetria.** A integral da parte positiva da variável diverge, assim como a da parte negativa em valor absoluto. Por exemplo,

$$
\int_0^R\frac{u}{\pi(1+u^2)}\,du
=
\frac{\log(1+R^2)}{2\pi}
\longrightarrow\infty.
$$

A esperança usual não existe: não podemos cancelar duas quantidades infinitas. A estacionariedade estrita não impediu esse comportamento, pois sua definição não exige momentos.

### Exemplo 3 — Fracamente estacionário, mas não estritamente estacionário

Considere variáveis independentes $\{Z_t:t\in\mathbb Z\}$ com distribuições alternadas:

- Se $t$ é par, $Z_t$ assume $-1$ e $1$, com probabilidade $1/2$ cada.
- Se $t$ é ímpar, $Z_t$ assume $-\sqrt2$, $0$ e $\sqrt2$, com probabilidades $1/4$, $1/2$ e $1/4$.

**Passo 1 — Média nos instantes pares.**

$$
E[Z_t]=-\frac12+\frac12=0.
$$

**Passo 2 — Média nos instantes ímpares.**

$$
E[Z_t]
=
-\sqrt2\frac14+0\frac12+\sqrt2\frac14
=
0.
$$

**Passo 3 — Segundo momento.**

Nos instantes pares,

$$
E[Z_t^2]=1\frac12+1\frac12=1.
$$

Nos ímpares,

$$
E[Z_t^2]=2\frac14+0\frac12+2\frac14=1.
$$

A média é zero em todos os instantes, logo a variância é sempre 1.

**Passo 4 — Covariâncias.** Para instantes distintos, a independência dá covariância zero. Assim,

$$
\operatorname{Cov}(Z_s,Z_t)
=
\begin{cases}
1,&s=t,\\
0,&s\ne t,
\end{cases}
$$

que depende apenas de $t-s$. O processo é fracamente estacionário.

**Passo 5 — Contradizer a estacionariedade estrita.** Basta comparar as marginais:

$$
P(Z_0=0)=0,
\qquad
P(Z_1=0)=\frac12.
$$

As distribuições mudam após um deslocamento de uma unidade. Falha a condição da estacionariedade estrita já para $n=1$.

Este exemplo mostra que invariância de média e covariância não garante invariância da distribuição.

### Exemplo 4 — Marginais iguais, média e variância constantes, mas nenhuma das duas formas

Sejam $\{U_k:k\in\mathbb Z\}$ variáveis independentes que assumem $-1$ e $1$ com probabilidades iguais. Construa pares consecutivos repetindo cada variável:

$$
Z_{2k}=U_k,
\qquad
Z_{2k+1}=U_k.
$$

Por exemplo,

$$
Z_0=U_0,\quad Z_1=U_0,\quad Z_2=U_1,\quad Z_3=U_1.
$$

**Passo 1 — Marginais e momentos individuais.** Todo $Z_t$ assume $-1$ e $1$ com as mesmas probabilidades. Portanto,

$$
E[Z_t]=0,
\qquad
\operatorname{Var}(Z_t)=1.
$$

**Passo 2 — Comparar pares com a mesma defasagem.** Os instantes 0 e 1 usam a mesma variável:

$$
\operatorname{Cov}(Z_0,Z_1)
=
E[U_0^2]=1.
$$

Os instantes 1 e 2 usam variáveis independentes, com médias zero:

$$
\operatorname{Cov}(Z_1,Z_2)
=
E[U_0U_1]
=
E[U_0]E[U_1]=0.
$$

Ambos os pares têm defasagem 1, mas suas covariâncias diferem. Falha a terceira condição da estacionariedade fraca.

**Passo 3 — Comparar as distribuições dos pares.**

$$
P(Z_0=1,Z_1=1)=P(U_0=1)=\frac12,
$$

enquanto

$$
P(Z_1=1,Z_2=1)
=
P(U_0=1,U_1=1)=\frac14.
$$

O segundo par é o primeiro deslocado em uma unidade. As probabilidades conjuntas diferem, portanto o processo também não é estritamente estacionário.

Não basta verificar que todas as distribuições individuais são iguais: a origem do relógio pode alterar a dependência entre instantes.

## 6. Comparação direta

| Questão | Estrita | Fraca |
|---|---|---|
| O que deve permanecer invariável? | Todas as distribuições finito-dimensionais sob deslocamento comum | Média e covariâncias sob deslocamento comum |
| Exige segundo momento finito? | Não | Sim |
| Exige marginais iguais em todos os instantes? | Sim | Não, em geral |
| Exige independência? | Não | Não |
| Exige trajetória constante? | Não | Não |
| Implica a outra definição? | Sim, se houver segundo momento finito | Não em geral; sim para processos gaussianos |

O termo “fraca” se refere ao tipo de informação preservada, os dois primeiros momentos. Não elimina a hipótese de finitude que a definição estrita não exige. Por isso, sem hipóteses adicionais, não há implicação geral em nenhum dos sentidos.

## 7. Três exercícios selecionados

**1. Constância de uma trajetória e propriedades do processo.** Seja $U$ com $P(U=-2)=P(U=2)=1/2$, e defina $Z_t=U$ para todo $t$. Demonstre diretamente a estacionariedade estrita e calcule média, variância e autocovariância. Explique por que a dependência perfeita não impede estacionariedade.

**2. Uma média constante não resolve tudo.** Seja $U$ com probabilidades iguais nos valores $-1$ e $1$, e defina $Z_t=tU$, $t\in\mathbb Z$. Calcule $E[Z_t]$, $\operatorname{Var}(Z_t)$ e $\operatorname{Cov}(Z_s,Z_t)$. Verifique as duas definições.

**3. Verificação incompleta.** Um colega examinou um processo e encontrou marginais iguais, todas com média zero e variância 1. Ele concluiu que o processo é estritamente e fracamente estacionário. Use o Exemplo 4 para refutar cada conclusão, indicando exatamente a condição que não foi verificada.

### Respostas para conferência

**1.** Todo vetor é $(U,\ldots,U)$ antes e depois do deslocamento. A média é zero e, como $U^2=4$, a variância e a autocovariância são 4 em todos os instantes e defasagens. O processo satisfaz ambas as definições.

**2.** Como $E[U]=0$ e $U^2=1$,

$$
E[Z_t]=0,\qquad
\operatorname{Var}(Z_t)=t^2,\qquad
\operatorname{Cov}(Z_s,Z_t)=st.
$$

A variância muda com o instante, então não há estacionariedade fraca. Também não há estrita: $Z_0=0$ com probabilidade 1, enquanto $Z_1$ assume somente $-1$ ou $1$.

**3.** Para estacionariedade fraca, faltou verificar se covariâncias dependem apenas da defasagem. Para estrita, faltou verificar todas as distribuições conjuntas; no contraexemplo, os pares já são suficientes para refutar a conclusão.
